# Strings
- A sequence of characters
- Inbuilt methods --> `upper()`, `lower()`, `replace()`, `split()`, `len`, `strip()`

```python
greeting = "Hello, Data Science!"
```

```python
# Upper case
greeting.upper()
```

```python
# Lower case
greeting.lower()
```

```python
# Replace()
greeting.replace("Hello", "Welcome to")
```

```python
# Split()
greeting.split()
```

```python
# Length of a string
len(greeting)
```

```python
# Strip()
print("      padded     ") # with spaces
print("   padded      word   ".strip())
print("    padded     ".strip())
```

```python
text = "This    is a cat"
print(text)
```

```python
text = text.split()
print(text)
```

```python
text.strip()
```

# f-strings

```python
first_name = "Ada"
last_name = "Lovelace"

print(first_name +" " + last_name)
```

```python
# f-string
print(f"{first_name} {last_name}")
```

```python
price = 19.9999
print(f"Price: ${price:.2f}") # :2f formats to 2dp
```
