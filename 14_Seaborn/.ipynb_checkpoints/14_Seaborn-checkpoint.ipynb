# Topic 14: Seaborn – Statistical Visualization

Seaborn is built on Matplotlib and creates beautiful statistical charts with less code.

pip install seaborn

````python
%matplotlib inline
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
np.random.seed(42)
sns.set_theme(style='whitegrid', palette='muted')

n = 300
df = pd.DataFrame({
    'Gender':     np.random.choice(['Male','Female'], n),
    'Department': np.random.choice(['Engineering','Sales','HR','Finance'], n),
    'Education':  np.random.choice(['BSc','MSc','PhD'], n),
    'Experience': np.random.randint(0, 30, n),
    'Salary':     np.random.normal(65000,18000,n).clip(25000,130000).round(-2),
    'Score':      np.random.normal(72, 12, n).clip(30,100).round(1),
    'Performance':np.random.choice([1,2,3,4,5], n),
})
df['Salary'] = (df['Salary'] + df['Experience']*1200).clip(25000,140000)
print('Dataset ready:', df.shape)
````

## 1. Distribution Plots


````python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(df["Salary"], bins=25, kde=True, ax=axes[0], color="steelblue")
axes[0].set_title("Salary Distribution")

for g, c in zip(["Male","Female"],["steelblue","coral"]):
    sns.kdeplot(df[df["Gender"]==g]["Salary"], ax=axes[1], fill=True, label=g, alpha=0.4, color=c)
axes[1].set_title("Salary by Gender"); axes[1].legend()

plt.tight_layout(); plt.show()
````

## 2. Box & Violin Plots

- **Box plot:** shows median, Q1, Q3, whiskers, and outliers.
- **Violin plot:** box plot + density curve — shows shape of distribution.


````python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.boxplot(data=df, x="Department", y="Salary", ax=axes[0], palette="Set2")
axes[0].set_title("Box Plot"); axes[0].tick_params(axis="x", rotation=15)

sns.violinplot(data=df, x="Department", y="Salary", ax=axes[1], palette="Set3", inner="quartile")
axes[1].set_title("Violin Plot"); axes[1].tick_params(axis="x", rotation=15)

plt.tight_layout(); plt.show()
````

## 3. Heatmaps

Correlation heatmaps show which variables are related.


````python
fig, axes = plt.subplots(1, 2, figsize=(13, 4))

corr = df[["Experience","Salary","Score","Performance"]].corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0, ax=axes[0])
axes[0].set_title("Correlation Heatmap")

pivot = df.pivot_table("Salary","Department","Education","mean")
sns.heatmap(pivot, annot=True, fmt=".0f", cmap="YlOrRd", ax=axes[1])
axes[1].set_title("Mean Salary by Dept & Education")

plt.tight_layout(); plt.show()
````

## 4. Pair Plot

Pair plots show scatter plots for every pair of numeric variables.


````python
sample = df[["Experience","Salary","Score","Gender"]].sample(100, random_state=1)
sns.pairplot(sample, hue="Gender",
             palette={"Male":"steelblue","Female":"coral"},
             diag_kind="kde", plot_kws={"alpha":0.5,"s":20})
plt.suptitle("Pair Plot", y=1.02); plt.show()
````

## 5. Scatter & Regression Plots


````python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.scatterplot(data=df, x="Experience", y="Salary",
                hue="Department", alpha=0.6, ax=axes[0])
axes[0].set_title("Experience vs Salary by Department")
axes[0].legend(fontsize=7)

sns.regplot(data=df, x="Experience", y="Salary", ax=axes[1],
            scatter_kws={"alpha":0.3}, line_kws={"color":"red","linewidth":2})
axes[1].set_title("With Regression Line")

plt.tight_layout(); plt.show()
````

## ✅ Summary

| Chart | Seaborn function |
|---|---|
| Histogram + density | `sns.histplot(kde=True)` |
| Density only | `sns.kdeplot(fill=True)` |
| Box plot | `sns.boxplot()` |
| Violin plot | `sns.violinplot()` |
| Heatmap | `sns.heatmap(annot=True)` |
| Pair plot | `sns.pairplot(hue=...)` |
| Scatter + trend | `sns.regplot()` |
