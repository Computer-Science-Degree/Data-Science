# Arithmetic Operators
#### These are mathematical operators 

- Addition +
- Subtraction -
- Multiplication *
- Division /
- Floor Division //
- Modulus or Remainder %
- Power ** or Exponentiation


```python
# Addition
x = 45
y = 15
result = x + y
print(result)
```

```python
# Subtraction 
diff = x - y
print(diff)
```

```python
# Multiplication - Product
price = 500
quantity = 4

total = price * quantity
print(total)
```

```python
# Division - Quotient
total_sales = 100000
number_of_customers = 200

average_sales = total_sales / number_of_customers
print(average_sales)
```

```python
# Floor Division - Returns the whole number
result = 10 // 3
result
```

```python
15 // 2
```

```python
# Modulus - Remainder
result = 10 % 3
result
```

```python
17 % 3
```

```python
# Exponentiation
result = 2 ** 3
result
```

```python
# Combine multiple operators to one expression
result = 10 + 5 * 2
result
```

```python
result = (10 + 5) * 2
result
```

    - not equal to `!=`

```python
#Equal to
10 == 10 # True
```

```python
11 == 3 # False
```
