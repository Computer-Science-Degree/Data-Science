# Topic 10: Understanding Data

**Topics:** Types of data · Descriptive statistics · Central tendency · Dispersion · Skewness & Kurtosis


## 1. Types of Data

| Category | Type | Examples |
|---|---|---|
| **Numerical** | Continuous | Height, salary, temperature |
| **Numerical** | Discrete | Goals scored, number of children |
| **Categorical** | Nominal | City, gender, blood type |
| **Categorical** | Ordinal | Education level, star ratings |


```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name":       ["Alice","Bob","Charlie"],
    "Age":        [25, 30, 22],          # Discrete numeric
    "Height":     [165.2, 178.5, 171.0], # Continuous numeric
    "City":       ["Accra","Kumasi","Tamale"],  # Nominal
    "Education":  ["BSc","MSc","BSc"],   # Ordinal
    "Rating":     [3, 5, 2],            # Ordinal (scale)
})
print(df)
print("\nData types:")
print(df.dtypes)
```

## 2. Descriptive Statistics


```python
import numpy as np, pandas as pd
np.random.seed(42)
df = pd.DataFrame({
    "Score":  np.random.normal(70,15,100).clip(0,100).round(1),
    "Salary": np.random.normal(60000,15000,100).clip(25000,120000).round(-2),
})
print(df.describe().round(2))
```

## 3. Measures of Central Tendency


```python
scores = df["Score"]
print(f"Mean   (average)    : {scores.mean():.2f}")
print(f"Median (middle)     : {scores.median():.2f}")
print(f"Mode   (most common): {scores.mode()[0]:.2f}")
print("""
When to use:
  Mean   → symmetric data, no outliers
  Median → skewed data or outliers (e.g. income)
  Mode   → categorical data
""")
```

## 4. Measures of Dispersion


```python
print(f"Range     : {scores.max()-scores.min():.2f}")
print(f"Std Dev   : {scores.std():.2f}")
print(f"Variance  : {scores.var():.2f}")
print(f"IQR       : {scores.quantile(0.75)-scores.quantile(0.25):.2f}")
print(f"Q1 (25%): {scores.quantile(0.25):.2f}")
print(f"Q2 (50%): {scores.quantile(0.50):.2f}")
print(f"Q3 (75%): {scores.quantile(0.75):.2f}")
```

## 5. Skewness and Kurtosis

- **Skewness = 0** → symmetric · **> 0** → right-skewed · **< 0** → left-skewed  
- **Kurtosis > 0** → heavy tails (more outliers) · **< 0** → light tails


```python
import numpy as np, pandas as pd
symm  = pd.Series(np.random.normal(50, 10, 1000))
right = pd.Series(np.random.exponential(10, 1000))
left  = pd.Series(100 - np.random.exponential(10, 1000))

print("Skewness:")
print(f"  Symmetric   : {symm.skew():.3f}")
print(f"  Right-skewed: {right.skew():.3f}")
print(f"  Left-skewed : {left.skew():.3f}")

print("\nKurtosis:")
print(f"  Symmetric   : {symm.kurtosis():.3f}")
```

## ✅ Summary

| Metric | Measures | Function |
|---|---|---|
| Mean | Center (average) | `.mean()` |
| Median | Center (robust) | `.median()` |
| Std Dev | Spread | `.std()` |
| IQR | Spread (robust) | `.quantile(0.75) - .quantile(0.25)` |
| Skewness | Asymmetry | `.skew()` |
| Kurtosis | Tail weight | `.kurtosis()` |
