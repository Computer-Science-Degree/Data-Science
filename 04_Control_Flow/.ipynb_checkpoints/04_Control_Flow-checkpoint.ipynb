```python
# Continue
for number in range(5):
    if number == 2:
        continue
    print(number)
```

# Exception Handling Control Flow
- try
- except
- finally

```python
# Example
try:
    number = int(input("Enter number: "))
    print(number)

except ValueError:
    print('Invalid Input')
```

```python
# Finally - always runs
try:
    print("Trying")
finally:
    print("Always Runs")
```

## 3. For Loop

Iterates over any sequence (list, string, range).


```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)

# range(start, stop, step)
for i in range(1, 11, 2):
    print(i, end=" ")  # 1 3 5 7 9
print()  # newline

# enumerate() gives index AND value
for idx, fruit in enumerate(fruits, start=1):
    print(f"{idx}. {fruit}")

# Loop over dictionary
person = {"name": "Alice", "age": 25, "city": "Accra"}
for key, value in person.items():
    print(f"  {key}: {value}")
```

## 4. While Loop

Runs as long as the condition is `True`.


```python
count = 0
while count < 5:
    print(f"Count: {count}")
    count += 1  # ALWAYS update the variable — prevents infinite loop

print("Done!")
```

## 5. Loop Control: break, continue, pass


```python
# break: stop the loop immediately
for i in range(10):
    if i == 5:
        print("Found 5! Stopping.")
        break
    print(i, end=" ")

print()

# continue: skip this iteration, continue to next
for i in range(10):
    if i % 2 == 0:
        continue       # skip even numbers
    print(i, end=" ")  # only prints odd numbers

print()

# pass: do nothing (placeholder)
for i in range(3):
    pass  # fill in later
```

## 6. Nested Loops


```python
# Multiplication table 1–5
for i in range(1, 6):
    for j in range(1, 6):
        print(f"{i*j:3}", end="")
    print()  # new line after each row
```

## 7. Real-World Example: Analysing Scores


```python
scores = [85, 42, 91, 67, 55, 78, 39, 95, 60, 73]

passed, failed = [], []
for score in scores:
    if score >= 50:
        passed.append(score)
    else:
        failed.append(score)

print(f"Passed : {len(passed)} — {passed}")
print(f"Failed : {len(failed)} — {failed}")
print(f"Average: {sum(scores)/len(scores):.1f}")
```

## ✅ Summary

| `for` | Iterate a known number of times |
| `while` | Repeat until a condition changes |
| `break` | Exit loop early |
| `continue` | Skip to next iteration |
| `pass` | Placeholder — do nothing |
