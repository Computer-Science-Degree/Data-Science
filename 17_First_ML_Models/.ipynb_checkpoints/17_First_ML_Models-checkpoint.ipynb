# Topic 17: First ML Models

We'll train 4 beginner models using scikit-learn.

**Models:** Linear Regression · Logistic Regression · KNN · Decision Tree


```python
%matplotlib inline
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report, mean_absolute_error, r2_score
import numpy as np; np.random.seed(42)

# Regression dataset
exp = np.random.randint(0,30,200).reshape(-1,1)
sal = 30000 + exp.ravel()*2500 + np.random.normal(0,6000,200)

# Classification dataset
n=200
X_clf = pd.DataFrame({"Age":np.random.randint(22,55,n),"Experience":np.random.randint(0,25,n),
                       "Salary":np.random.normal(60000,15000,n).clip(25000,120000),
                       "Score":np.random.normal(72,12,n).clip(30,100)})
y_clf = (X_clf["Salary"]>65000).astype(int)

X_tr,X_te,y_tr,y_te = train_test_split(X_clf,y_clf,test_size=0.2,random_state=42)
sc = StandardScaler()
X_tr_s = sc.fit_transform(X_tr); X_te_s = sc.transform(X_te)
print("Datasets ready.")
```

## 1. Linear Regression

> Fits the best straight line through the data to predict a **continuous number**.


```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error
import numpy as np

e_tr, e_te, s_tr, s_te = train_test_split(exp, sal, test_size=0.2, random_state=42)
model_lr = LinearRegression()
model_lr.fit(e_tr, s_tr)

y_pred = model_lr.predict(e_te)
print(f"Equation: Salary = {model_lr.coef_[0]:,.0f} x Experience + {model_lr.intercept_:,.0f}")
print(f"MAE: ${mean_absolute_error(s_te, y_pred):,.0f}")
print(f"R² : {r2_score(s_te, y_pred):.3f}")

for h in [2, 10, 20]:
    print(f"  {h} years → ${model_lr.predict([[h]])[0]:,.0f}")
```

## 2. Logistic Regression

> Predicts a **probability** and classifies into 0 or 1.


```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

model_log = LogisticRegression(random_state=42)
model_log.fit(X_tr_s, y_tr)
y_pred_log = model_log.predict(X_te_s)

print(f"Accuracy: {accuracy_score(y_te, y_pred_log):.2%}")
print(classification_report(y_te, y_pred_log, target_names=["Class 0","Class 1"]))
```

## 3. K-Nearest Neighbors (KNN)

> Classifies based on the **k most similar** training examples.


```python
from sklearn.neighbors import KNeighborsClassifier

print(f"Trying different k values:")
for k in [1, 3, 5, 7, 11, 15]:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_tr_s, y_tr)
    acc = accuracy_score(y_te, knn.predict(X_te_s))
    print(f"  k={k:>2}: {acc:.2%}")
```

## 4. Decision Tree

> Makes predictions by asking a series of **yes/no questions**.


```python
from sklearn.tree import DecisionTreeClassifier, export_text

model_dt = DecisionTreeClassifier(max_depth=4, random_state=42)
model_dt.fit(X_tr, y_tr)  # no scaling needed for trees
print(f"Accuracy: {accuracy_score(y_te, model_dt.predict(X_te)):.2%}")
print("\nTree rules:")
print(export_text(model_dt, feature_names=list(X_clf.columns)))
```

## 5. Model Comparison


```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

results = {
    "Logistic Regression": accuracy_score(y_te, LogisticRegression(random_state=42).fit(X_tr_s,y_tr).predict(X_te_s)),
    "KNN (k=5)":           accuracy_score(y_te, KNeighborsClassifier(5).fit(X_tr_s,y_tr).predict(X_te_s)),
    "Decision Tree (d=4)": accuracy_score(y_te, DecisionTreeClassifier(max_depth=4,random_state=42).fit(X_tr,y_tr).predict(X_te)),
}
print(f"{"Model":<25} {"Accuracy":>10}")
print("-"*38)
for name, acc in sorted(results.items(), key=lambda x:-x[1]):
    print(f"{name:<25} {acc:>10.2%}  {"█"*int(acc*30)}")
```
